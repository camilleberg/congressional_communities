```python
def make_line_comp_chart(age_group_label):
    # now making the little comparison chart 
    fig = plt.figure(figsize=(8, 2))
    ax = fig.add_subplot(111)

    # color blocking
    not_used = "darkgrey"
    color_used = okabe_ito_colors[age_bracket_cols.index(age_group_label)]

    # to make numberline
    intervals = np.round(np.linspace(0, 1, 11), 2)
    intervals_y = np.repeat(1, 11)
    plt.plot(intervals, intervals_y, marker='|',color='lightgrey', linestyle='solid',
        linewidth=1, markersize=10, alpha = 0.5, 
        zorder=1) # this is for directional order

    # calculating the % of each group 
    x_cc =  df_state_graph[ccn20_record['State']][age_group_label] / df_state_graph[ccn20_record['State']]['Total Population']
    x_cd = df_cd_graph[ccn20_record['DC']][age_group_label] / df_cd_graph[ccn20_record['DC']]['Total Population'] 
    x_state = df_cc_graph[age_group_label] / df_cc_graph['Total Population']

    # adding plots
    plt.scatter(x = x_state, 
                y = 1, color = not_used, s = 150, 
                edgecolors='white', linewidths=1,
                hatch='/////', # hatching
                zorder=2, # for order
                label = 'State')
    plt.scatter(x = x_cd, 
                y = 1, color = not_used, s = 150, 
                edgecolors='white', linewidths=1,
                zorder=3, # for order
                label = 'Congressional District')
    plt.scatter(x = x_cc, 
                y = 1, color = color_used, s= 150, 
                edgecolors='white', linewidths=1,
                zorder=4, # for order
                label = 'Congressional Community')
    plt.axis('off')

    # labeling axis thing
    for x in intervals:
        ax.annotate('%s' % x, xy=(x, 0.97), textcoords='data')
    
    # dynaically adjusting width
    x_min = floor_to_decimals(min(x_cc, x_cd, x_state), 1) - 0.05
    x_max = ceil_to_decimals(max(x_cc, x_cd, x_state), 2) + 0.1
    plt.xlim(x_min, x_max) 
    
    # layout
    plt.legend(ncol=3, loc="best") 
    plt.tight_layout()

    
    # plt.savefig(f'./figures/age_comparisons_line_{age_group_label}.png')
    plt.show()
    
    return plt
```
